## 2.Firm-level Climate Scores

### 2.1 Sautner et al. (2023)

```python
import pandas as pd
import numpy as np

sautner = pd.read_excel("firmquarter_score_2024Q4_Version_2025_Jul_03.xlsx")
sautner = sautner[sautner["hqcountrycode"] == "US"].copy()
sautner.head(10)
```

```python
# repair 'gvkey'
sautner["gvkey"] = (
    pd.to_numeric(sautner["gvkey"], errors="coerce")
      .astype("Int64")
)
```

```python
# link finvar.xlsx
fin = pd.read_excel("finvar.xlsx")
fin.head(10)
```

```python
# link finvar.xlsx
fin = pd.read_excel("finvar.xlsx")
fin["datadate"] = pd.to_datetime(fin["datadate"])
fin["year"] = fin["datadate"].dt.year
fin["quarter"] = fin["datadate"].dt.quarter

sautnerfin = fin.merge(
    sautner,
    on=['gvkey', 'year', 'quarter'],
    how='inner'
)
```

```python
print(len(fin))
print(len(sautner))
print(len(sautnerfin))
```

### 2.2 LSTY(2024)

```python
lsty = pd.read_excel("measures_2025.xlsx")
lsty.head(10)
```

```python
# repair "qtr"
lsty = lsty.rename(
    columns={
        "qtr": "quarter"
    }
)
```

```python
dup_lsty = lsty[
    lsty.duplicated(
        subset=["gvkey","year","quarter"],
        keep=False
    )
].sort_values(
    ["gvkey","year","quarter"]
)

dup_lsty.head(20)
```

```python
dup_lsty.groupby(
    ["gvkey","year","quarter"]
)["tran_risk_w"].nunique().describe()
```

```python
# We should deal with REPLICATION!
# lsty
target_cols_lsty = lsty.select_dtypes(include=np.number).columns.tolist()

for c in ["gvkey", "year", "quarter"]:
    if c in target_cols_lsty:
        target_cols_lsty.remove(c)

# take means
lsty_avg = (
    lsty
    .groupby(["gvkey", "year", "quarter"], as_index=False)
    [target_cols_lsty]
    .mean()
)

print(len(lsty))
print(len(lsty_avg))

# examine again
print(
    lsty_avg.duplicated(
        subset=["gvkey","year","quarter"]
    ).sum()
)
```

```python
# sautnerfin
# delete some vars
delete_var = ['cusip_x', 'isin', 'cusip_y', 'hqcountrycode']
sautnerfin = sautnerfin.drop(
    columns=delete_var,
    errors='ignore'
)
```

```python
agg_dict = {}

for col in sautnerfin.columns:

    if col in ["gvkey", "year", "quarter"]:
        continue

    if pd.api.types.is_numeric_dtype(sautnerfin[col]):
        agg_dict[col] = "mean"
    else:
        agg_dict[col] = "first"

sautnerfin_avg = (
    sautnerfin
    .groupby(
        ["gvkey", "year", "quarter"],
        as_index=False
    )
    .agg(agg_dict)
)

print(
    sautnerfin_avg.duplicated(
        subset=["gvkey","year","quarter"]
    ).sum()
)
```

```python
# link together
lstysautnerfin = sautnerfin_avg.merge(
    lsty_avg,
    on=["gvkey","year","quarter"],
    how="inner"
)
```

```python
print(len(fin))
print(len(sautner))
print(len(sautnerfin))
print(len(lsty))
print(len(lstysautnerfin))
```
