## 1.Financial Variables

```python
# import the data downloaded from CompuStat
import pandas as pd
import numpy as np

df = pd.read_excel("reprawdata.xlsx")
```

```python
df.head(10)
```

```python
# delete samples in finance industry
df = df[
    (df['sic'] < 6000)
    | (df['sic'] > 6999)
]
```

```python
# manage dates
df['datadate'] = pd.to_datetime(df['datadate'])
df = df.sort_values(['gvkey', 'datadate'])
df.head(10)
```

```python
# Tobin's q
df['mve'] = df['prccq'] * df['cshoq']

df['tobinq'] = (
    df['mve']
    + df['atq']
    - df['ceqq']
) / df['atq']
```

```python
# Controls
df['log_assets'] = np.log(df['atq'])

df['capex'] = df['capxy'] / df['atq']

df['ppe'] = df['ppentq'] / df['atq']

df['book_leverage'] = (
    df['dlttq'] + df['dlcq']
) / df['atq']

df['roa'] = df['ibq'] / df['atq']
```

```python
# take lags
controls = [
    'log_assets',
    'capex',
    'ppe',
    'book_leverage',
    'roa'
]

for v in controls:
    df[v+'_lag'] = df.groupby('gvkey')[v].shift(1)
```

```python
# industry*quarter fe
df['quarter'] = df['datadate'].dt.to_period('Q')
df['naics2'] = (
    df['naics']
      .astype(str)
      .str[:2]
)

df['ind_quar_fe'] = (
    df['naics2'].astype(str)
    + "_"
    + df['quarter'].astype(str)
)
df.head(10)
```

```python
# output: finvar.xlsx
keep_cols = [
    'gvkey',
    'datadate',
    'cusip',
    'naics',
    'sic',

    'tobinq',
    
    'log_assets_lag',
    'capex_lag',
    'ppe_lag',
    'book_leverage_lag',
    'roa_lag',

    'ind_quar_fe'
]
df_fin = df[keep_cols]
df_fin.to_excel(
    "finvar.xlsx",
    index=False
)
```
